```python
import pandas as pd
df = pd.read_csv("Traffic.csv")
df['Traffic Situation'] = df['Traffic Situation'].replace({
    'heavy': 'high',
    'normal': 'medium'
})
label_map = {'low': 0, 'medium': 1, 'high': 2}
df['label'] = df['Traffic Situation'].map(label_map)
df['Hour'] = pd.to_datetime(df['Time'], errors='coerce').dt.hour
df.drop(columns=['Time', 'Date', 'Day of the week', 'Traffic Situation', 'Total'], inplace=True)
df = df[['Hour', 'CarCount', 'BikeCount', 'BusCount', 'TruckCount', 'label']]
print(df.head())
```

```python
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import MinMaxScaler
import numpy as np
X = df.drop('label', axis=1).values
y = df['label'].values
scaler = MinMaxScaler()
X_scaled = scaler.fit_transform(X)
X_reshaped = X_scaled.reshape((X_scaled.shape[0], 1, X_scaled.shape[1]))
X_train, X_test, y_train, y_test = train_test_split(X_reshaped, y, test_size=0.2, random_state=42)
print("Train shape:", X_train.shape)
print("Test shape:", X_test.shape)
```

```python
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import LSTM, Dense
from tensorflow.keras.callbacks import EarlyStopping
model = Sequential()
model.add(LSTM(64, input_shape=(1, X_train.shape[2])))
model.add(Dense(32, activation='relu'))
model.add(Dense(3, activation='softmax'))
model.compile(optimizer='adam', loss='sparse_categorical_crossentropy', metrics=['accuracy'])
early_stop = EarlyStopping(monitor='val_loss', patience=5, restore_best_weights=True)
history = model.fit(X_train, y_train, epochs=50, batch_size=32,
                    validation_split=0.2, callbacks=[early_stop])
```

```python
from sklearn.metrics import classification_report
from sklearn.preprocessing import LabelEncoder
import numpy as np
y_pred_probs = model.predict(X_test)
y_pred = np.argmax(y_pred_probs, axis=1)
label_encoder = LabelEncoder()
label_encoder.fit(['low', 'medium', 'high']) 
y_test_labels = label_encoder.inverse_transform(y_test)
y_pred_labels = label_encoder.inverse_transform(y_pred)
print(classification_report(y_test_labels, y_pred_labels))
```

```python
def suggest_vehicle(traffic_level):
    suggestions = {
        'low': 'Any vehicle can be used (Car, Bike, Bus)',
        'medium': 'Prefer Bike or Carpooling',
        'high': 'Prefer Two-wheelers or Delay Travel if possible'
    }
    return suggestions.get(traffic_level.lower(), "No suggestion")
for true, pred in zip(y_test_labels[:5], y_pred_labels[:5]):
    print(f"Predicted: {pred}, Suggestion: {suggest_vehicle(pred)}")
```

```python
import pandas as pd
predicted_labels = label_encoder.inverse_transform(y_pred)
test_df = pd.DataFrame(X_test.reshape(X_test.shape[0], -1))  
test_df['Actual'] = label_encoder.inverse_transform(y_test)
test_df['Predicted'] = predicted_labels
test_df['Suggestion'] = test_df['Predicted'].apply(suggest_vehicle)
test_df.to_csv('test_predictions.csv', index=False)
print("✅ Saved predictions to test_predictions.csv")
```

```python
from ultralytics import YOLO
import cv2
from collections import defaultdict
model = YOLO('yolov8n.pt')  
video_path = 'C:/Users/Main Python2/Downloads/10.mp4' 
cap = cv2.VideoCapture(video_path)
vehicle_classes = ['car', 'motorcycle', 'bus', 'truck']
frame_counts = []
while True:
    ret, frame = cap.read()
    if not ret:
        break
    results = model.predict(frame, conf=0.3, verbose=False)[0]
    vehicle_count = defaultdict(int)
    for box in results.boxes:
        cls_id = int(box.cls[0])
        class_name = model.names[cls_id]
        if class_name in vehicle_classes:
            vehicle_count[class_name] += 1
    frame_counts.append(vehicle_count)
    annotated = results.plot()
    cv2.imshow("YOLOv8 Detection", annotated)
    if cv2.waitKey(1) & 0xFF == ord('q'):
        break
cap.release()
cv2.destroyAllWindows()
final_counts = defaultdict(int)
for fc in frame_counts:
    for k, v in fc.items():
        final_counts[k] += v
print("\n🚗 Total Vehicle Counts from Processing Video:")
for vehicle, count in final_counts.items():
    print(f"{vehicle}: {count}")
```
